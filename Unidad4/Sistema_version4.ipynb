```python
class Libro:
  def __init__(self,isbn,titulo,autor,categoria,precio):
    self.isbn=isbn
    self.titulo=titulo
    self.autor=autor
    self.categoria=categoria
    self.precio=precio
    self.prestado = False # Added to track if a specific book instance is lent

  def mostrar_info(self):
    estado ="prestado " if self.prestado else"Disponible"
    print(f"[{self.isbn}] {self.titulo} - {self.autor} | {self.categoria} | ${self.precio} | Estado: {estado}")

from datetime import datetime
from abc import ABC, abstractmethod

class Persona(ABC):
  def __init__(self, nombre, correo, fecha_nacimiento):
    self.nombre = nombre
    self.correo = correo
    self.fecha_nacimiento = fecha_nacimiento

  @abstractmethod
  def mostrar_info(self):
   """Metodo abstracto: las subclases deben redefinirlo"""
   pass

class Estudiante(Persona):
  def __init__(self,num_control,nombre,carrera,fecha_nacimiento,telefono,correo):
    super().__init__(nombre,correo,fecha_nacimiento)
    self.num_control=num_control
    self.carrera=carrera
    self.telefono=telefono
  def mostrar_info(self):
   return(f"[Estudiante] Nombre: {self.nombre}, Numero de control: {self.num_control}, Carrera: {self.carrera}, Telefono: {self.telefono}, Correo: {self.correo}, Fecha Nacimiento: {self.fecha_nacimiento}")


class Trabajador(Persona,Registrable):
  def __init__(self,num_trabajador, nombre, correo, fecha_nacimiento, departamento, usuario, contrasena): # Added departamento as a parameter
    super().__init__(nombre, correo, fecha_nacimiento)
    self.num_trabajador=num_trabajador
    self.departamento=departamento
    self.usuario=usuario
    self.contrasena=contrasena
    self.actividades = []  #Registro de acciones realisadas

  def mostrar_info(self):
    return (f"[Trabajador] Usuario: {self.usuario}, Numero Trabajador: {self.num_trabajador}, Nombre: {self.nombre}, Departamento: {self.departamento}")
  #Implementacion del metodo definido en la interfaz registrable
  def registrar_actividad(self, accion):
    fecha_actual = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    registro = f"El {fecha_actual}-{self.nombe} realizo: {accion}"
    self.actividades.append(registro)


class InventarioLibro:
  def __init__(self,libro,cantidad):
    self.libro=libro
    self.total=cantidad
    self.prestados=0
    self.vendidos=0

  def disponibles(self):
    return self.total-self.prestados-self.vendidos

  def prestar(self):
    if self.disponibles()>0:
       self.prestados+=1
       return True
    return False

  def devolver(self): # Corrected indentation
    if self.prestados>0:
      self.prestados-=1
      return True
    return False

  def vender(self):
    if self.disponibles()>0:
      self.vendidos+=1
      return True
    return False

  def mostrar_estado(self):
    l=self.libro # Changed self.l to self.libro
    print(f"[{l.isbn}] {l.titulo} - {l.autor} | Total: {self.total} \
      | Disponibles: {self.disponibles()} | Prestados: {self.prestados} | Vendidos: { self.vendidos}")
```

```python
class Prestamo:
  contador_prestamos=1

  def __init__(self,num_control,isbn,fecha_prestamo,fecha_devolucion):
    self.folio=Prestamo.contador_prestamos
    Prestamo.contador_prestamos+=1
    self.num_control=num_control
    self.isbn=isbn
    self.fecha_prestamo=fecha_prestamo
    self.fecha_devolucion=fecha_devolucion

  def mostrar_info(self):
    print(f"Folio: {self.folio} | Estudiante: {self.num_control} | Libro: {self.isbn} \
    Del {self.fecha_prestamo} al {self.fecha_devolucion}") # Fixed sel.isbn to self.isbn
```

```python
class Venta: # Renamed from 'venta' to 'Venta' for class naming convention
  contador_ventas=1 # Fixed cntador_ventas to contador_ventas
  def __init__(self,num_control,isbn,fecha_venta,cantidad,importe):
    self.folio_venta=Venta.contador_ventas
    Venta.contador_ventas+=1
    self.isbn=isbn
    self.num_control=num_control
    self.fecha_venta=fecha_venta
    self.cantidad=cantidad
    self.importe=importe

  def mostrar_info(self):
      print(f"Folio: {self.folio_venta} |  ISBN: {self.isbn} | Comprador: {self.num_control} | \
    Fecha {self.fecha_venta} | Cantidad {self.cantidad} | Importe {self.importe}") # Fixed sel.isbn to self.isbn
```

```python
def alta_libro():
  print("\n---Alta de Libro---") # Corrected escape sequence
  isbn=input("ISBN: ")
  for inv in inventarios: # Iterate over inventarios to check for existing books
    if inv.libro.isbn==isbn: # Check the isbn of the book object inside InventarioLibro
      print("Libro ya registrado")
      return
  titulo=input("Titulo: ")
  autor=input("Autor: ")
  categoria=input("Categoria: ")
  try:
    precio=float(input("Precio: "))
  except ValueError:
    print("Precio invalido")
    return
  try:
    cantidad=int(input("Cantidad de libros a agregar: ")) # Added quantity input
  except ValueError:
    print("Cantidad invalida")
    return
  nuevo_libro=Libro(isbn,titulo,autor,categoria,precio)
  inventario_item = InventarioLibro(nuevo_libro, cantidad)
  inventarios.append(inventario_item)
  print(f"Libro {titulo} agregado al inventario con {cantidad} unidades.") # More informative message
  trabajador_activo.registrar_actividad(f"Alta de libro'{titulo}'")

def alta_estudiante():
  print("\n---Alta de Estudiante---") # Corrected escape sequence
  num_control=input("Numero de control: ")
  for est in estudiantes:
    if est.num_control==num_control:
      print("Estudiante ya registrado")
      return
  nombre=input("Nombre: ")
  carrera=input("Carrera: ")
  fecha_nacimiento=input("Fecha de nacimiento (dd/mm/aaaa): ")
  telefono=input("Telefono: ")
  correo=input("Correo: ")
  nuevo_estudiante=Estudiante(num_control,nombre,carrera,fecha_nacimiento,telefono,correo)
  estudiantes.append(nuevo_estudiante)
  print(f"Estudiante {nombre} registrado con exito")
  trabajador_activo.registrar_actividad(f"Alta de estudiante'{nombre}'")


def consultar_libro():
  print("\n---Consultar Libro---") # Corrected escape sequence
  isbn=input("ingrese el ISBN del libro: ")
  found = False
  for inv_item in inventarios: # Iterate over inventarios to find the book
    if inv_item.libro.isbn==isbn:
      inv_item.mostrar_estado() # Show the state of the inventory item
      found = True
      return
  if not found:
    print( "Libro no encontrado")

def consultar_estudiante():
  print("\n---Consultar Estudiante---") # Corrected escape sequence
  num_control=input("ingrese el numero de control del estudiante: ")
  found = False
  for est in estudiantes:
    if est.num_control==num_control:
      print(est.mostrar_info()) # Print the returned string
      found = True
      return
  if not found:
    print( "Estudiante no encontrado")

def mostrar_inventario():
  print("\n---Inventario de todos los libros---") # Corrected escape sequence
  if not inventarios:
    print("Inventario vacio")
  else:
    for item in inventarios:
      item.mostrar_estado()

def mostrar_todos_estudiantes():
  print("\n---Estudiantes Registrados---") # Corrected escape sequence
  if not estudiantes:
    print("No hay estudiantes registrados")
  else:
    for est in estudiantes:
      print(est.mostrar_info()) # Print the returned string

def realizar_prestamo():
  print("\n---Realizar Prestamo---") # Corrected escape sequence
  num_control=input("Numero de control del estudiante: ")
  estudiante=next((e for e in estudiantes if e.num_control==num_control),None)
  if not estudiante:
    print("Estudiante no encontrado")
    return
  isbn=input("ISBN del libro: ")
  inventario_libro_item=next((i for i in inventarios if i.libro.isbn==isbn),None) # Renamed to avoid confusion with book object
  if not inventario_libro_item:
    print("Libro no encontrado en el inventario")
    return
  if not inventario_libro_item.prestar(): # Call prestar on the InventarioLibro object
    print("No hay copias disponibles para prestamo")
    return
  fecha_inicio=input("Fecha de inicio del prestamo (dd/mm/aaaa): ")
  fecha_termino=input("Fecha de termino del prestamo (dd/mm/aaaa): ")
  prestamo=Prestamo(num_control,isbn,fecha_inicio,fecha_termino)
  prestamos.append(prestamo)
  inventario_libro_item.libro.prestado=True # Mark the specific book as prestado
  print(f"Prestamo registrado con folio {prestamo.folio}.")
  trabajador_activo.registrar_actividad(f"prestamo realizado {num_control}")

def mostrar_todos_prestamos():
  print("\n---Prestamos Registrados---") # Corrected escape sequence
  if not prestamos:
    print("No hay prestamos registrados")
  else:
    for p in prestamos:
      p.mostrar_info()

def devolver_libro():
  print("\n---Devolver Libro---") # Corrected escape sequence
  try:
    folio=int(input("Folio del prestamo: "))
  except ValueError:
    print("Folio invalido")
    return
  prestamo_a_devolver=next((x for x in prestamos if x.folio==folio),None) # Renamed variable to avoid clash
  if not prestamo_a_devolver:
    print("Prestamo no encontrado")
    return

  isbn_libro_devuelto = prestamo_a_devolver.isbn
  inventario_item = next((i for i in inventarios if i.libro.isbn == isbn_libro_devuelto), None)

  if inventario_item:
    if inventario_item.devolver(): # Call devolver on the InventarioLibro object
      inventario_item.libro.prestado = False # Mark the specific book as not prestado
      prestamos.remove(prestamo_a_devolver)
      print(f"Libro con ISBN {isbn_libro_devuelto} devuelto con exito. Prestamo folio {folio} eliminado.")
    else:
      print("Error al devolver el libro: No hay prestamos registrados para este libro en el inventario.")
  else:
    print("Libro no encontrado en el inventario.")
```

```python
def registrar_venta():
  print("\n---Registrar Venta---") # Fixed escape sequence and indentation
  isbn = input("ISBN del libro a vender: ") # Added colon for consistency
  inventario_item = next((inv for inv in inventarios if inv.libro.isbn == isbn),None) # Use inventarios
  if not inventario_item:
    print("Libro no encontrado en el inventario.")
    return

  try:
    cantidad=int(input("Cantidad a vender: (solo 1 por el momento) ")) # Corrected typo, added colon
    if cantidad != 1 : # Ensure only 1 is sold
      print("Solo se permite vender una unidad por el momento.") # More informative message
      return
  except ValueError:
    print("Cantidad invalida.")
    return

  if not inventario_item.vender(): # Call vender on the InventarioLibro object
    print("No hay copias disponibles para la venta.")
    return

  num_control = input("Numero de control del comprador (999999 si es publico general): ") # Added colon

  if num_control != "999999":
   estudiante = next((e for e in estudiantes if e.num_control == num_control),None)
   if not estudiante:
     print("Estudiante no encontrado.")
     # Optional: If the sale should be reverted if student not found, add logic here.
     return
  fecha=input("Fecha de la venta (dd/mm/aaaa): ") # Added for consistency
  importe = inventario_item.libro.precio * cantidad # Use inventario_item.libro.precio
  venta = Venta(num_control,isbn,fecha,cantidad,importe) # Use Venta class
  ventas.append(venta)
  # libros.remove(libro) # This line should be removed, as we update inventory count, not remove the book
  print(f"Venta registrada. Folio: {venta.folio_venta}")
```

```python
def mostrar_historial_ventas():
  print("\n---Historial de Ventas---") # Corrected escape sequence
  if not ventas:
    print("No hay ventas registradas")
  else:
    for v in ventas:
      v.mostrar_info()
```

```python
def mostrar_menu():
  print("\n---SISTEMA DE LIBERACION ESCOLAR---") # Corrected escape sequence
  print("1. Dar de alta Libro")
  print("2. Dar de alta Estudiante")
  print("3. Consultar Libro por ISBN")
  print("4. Consultar Estudiante por numero de control")
  print("5. Mostrar inventario")
  print("6. Mostrar todos los estudiantes")
  print("7. Realizar prestamo")
  print("8. Mostrar todos los prestamos")
  print("9. Devolver libro")
  print("10. Registrar venta")
  print("11. Mostrar historial de ventas")
  print("12. Dar de alta trabajador")
  print("13. Mostrar trabajadores")
  print("14. Salir del sistema")
```

```python
def iniciar_sistema():
  sesion=iniciar_sesion()
  while sesion:
    mostrar_menu()
    opcion=input("Seleccione una opcion: ")
    if opcion=="1":
      alta_libro()
    elif opcion=="2":
      alta_estudiante()
    elif opcion=="3":
      consultar_libro()
    elif opcion=="4":
      consultar_estudiante()
    elif opcion=="5":
      mostrar_inventario()
    elif opcion=="6":
      mostrar_todos_estudiantes()
    elif opcion=="7":
      realizar_prestamo()
    elif opcion=="8":
      mostrar_todos_prestamos()
    elif opcion=="9":
      devolver_libro()
    elif opcion=="10":
      registrar_venta()
    elif opcion=="11":
      mostrar_historial_ventas()
    elif opcion=="12":
      alta_trabajador()
    elif opcion=="13":
      mostrar_trabajadores()
    elif opcion=="14":
      print("Saliendo del sistema. ¡Hasta luego!")
      break
    else:
      print("Opcion invalida. Por favor, seleccione una opcion valida.")
```

```python
inventarios=[]
estudiantes=[]
prestamos=[]
ventas=[]

iniciar_sistema()
```
