```python
class Producto:
  def __init__ (self, id, nombre, categoria, precio):
    self.id = id
    self.nombre = nombre
    self.categoria = categoria
    self.precio = precio

  def mostrar_info(self):
    print(f"[{self.id}] {self.nombre} - {self.categoria} | ${self.precio}")

class Cliente:
  def __init__ (self, id_cliente, nombre, telefono, correo):
    self.id_cliente = id_cliente
    self.nombre = nombre
    self.telefono = telefono
    self.correo = correo

  def mostrar_info(self):
    print(f"[{self.id_cliente} - {self.nombre} | Tel: {self.telefono} Cor: {self.correo}]")

class Venta:
  def __init__ (self, folio, id_cliente, codigo_producto, fecha, cantidad, total):
    self.folio = folio
    self.id_cliente = id_cliente
    self.codigo_producto = codigo_producto
    self.fecha = fecha
    self.cantidad = cantidad
    self.total = total

  def mostrar_info(self):
    print(f"[Se vendio {self.codigo_producto} con una cantidad de {self.total} al cliente con el siguiente ID {self.id_cliente} el dia {self.fecha}]")

class InventarioProducto:
  def __init__ (self, producto, cantidad, vendidos):
    self.producto = producto
    self.cantidad = cantidad
    self.vendidos = vendidos

  def disponibles(self):
    Disponibles = self.cantidad - self.vendidos
    print(f"Del producto {self.producto}, quedan disponibles: {Disponibles} copia(s)")

  def mostrar_estado(self):
    Disponibles = self.cantidad - self.vendidos
    print(f"[{self.producto.nombre} Cantidad: {Disponibles} Vendidos: {self.vendidos}]")

  def Agregar_Stock(self):
    Añadido = int(input("¿Cuantos productos desea agregar?: "))
    self.cantidad += Añadido



##### FUNCIONES #####

def alta_producto():
  print("\n --- ALTA DE PRODUCTO --- ")
  id = input("ID: ")
  for producto in productos:
    if producto.id == id:
      print("Ya existe un producto con ese ID")
      return
  nombre = input("Nombre: ")
  categoria = input("Categoria: ")
  try:
    precio = float(input("Precio: "))
  except ValueError:
    print("Precio invalido, use numeros decimales")
    return

  cantidad_inicial = int(input("Cantidad inicial en inventario: "))

  nuevo_producto = Producto(id, nombre, categoria, precio)
  productos.append(nuevo_producto)

  nuevo_inventario = InventarioProducto(nuevo_producto, cantidad_inicial, 0)
  inventarios.append(nuevo_inventario)
  print("Producto agregado con exito")

def Añadir_Inventario():
  print("\n --- AGREGAR STOCK A UN PRODUCTO --- ")
  id = input("ID del producto: ")
  for producto in productos:
    if producto.id == id:
      inventario = next((i for i in inventarios if i.producto.id == id), None)
      if not inventario:
        print("No hay inventario registrado para este producto")
        return
      inventario.Agregar_Stock()
      print("Stock agregado con exito")
      return

def Mostrar_Inventario():
  print("\n --- INVENTARIO --- ")
  if not inventarios:
    print("Inventario vacio")
  for item in inventarios:
    item.mostrar_estado()


def Consultar_Producto():
  print("\n --- CONSULTAR PRODUCTO POR CODIGO --- ")
  id = input("ID: ")
  for producto in productos:
    if producto.id == id:
      producto.mostrar_info()
      return
    else:
      print("No existe un producto con ese codigo.")


def Registrar_Cliente():
  print("\n --- REGISTRO DE CLIENTE --- ")
  id = input("ID: ")
  for cliente in clientes:
    if cliente.id_cliente == id:
      print("Ya existe un cliente con ese ID")
      return
  nombre = input("Nombre: ")
  telefono = input("Telefono: ")
  correo = input("Correo: ")
  nuevo_cliente = Cliente(id, nombre, telefono, correo)
  clientes.append(nuevo_cliente)
  print("Cliente registrado con exito.")

def Consultar_Cliente():
  print("\n --- CONSULTAR CLIENTE POR ID --- ")
  id_cliente = input("ID: ")
  for cliente in clientes:
    if cliente.id_cliente == id_cliente:
      cliente.mostrar_info()
      return
    else:
      print("No existe un cliente con ese ID")

def Mostrar_Clientes():
  print("\n --- CLIENTES REGISTRADOS --- ")
  if not clientes:
    print("No hay clientes registrados")
  for cliente in clientes:
    cliente.mostrar_info()

def Registrar_Venta():
  print("\n --- REGISTRO DE VENTA --- ")
  id = input("ID del producto a vender: ")
  producto = next((p for p in productos if p.id == id), None)
  if not producto:
    print("No existe un producto con ese ID")
    return
  try:
    cantidad = int(input("Cantidad a vender: "))
  except ValueError:
    print("Cantidad invalida")
    return
  id_cliente = input("ID del comprador (999999 si es publico general): ")
  if id_cliente != "999999":
    cliente = next((c for c in clientes if c.id_cliente == id_cliente), None)
    if not cliente:
      print("No existe un cliente con ese ID")
      return
    fecha = input("Fecha de la venta (dd/mm/aaaa): ")
    total = producto.precio * cantidad
    codigo_producto = id
    folio = len(ventas) + 1
    venta = Venta(folio, id_cliente, codigo_producto, fecha, cantidad, total)
    ventas.append(venta)
    ventas.append(cliente)
    print(f"Venta registrada. Folio: {folio}")

  inventario = next((i for i in inventarios if i.producto.id == id), None)
  if not inventario:
    print("No hay inventario registrado para este producto")
    return
  if inventario.cantidad - inventario.vendidos < cantidad:
    print("No hay suficiente stock disponible")
    return
  inventario.vendidos += cantidad
  inventario.cantidad - cantidad
  print("Venta registrada con exito")

def historial_ventas():
  print("\n --- HISTORIAL DE VENTAS --- ")
  if not ventas:
    print("No hay ventas registradas")
    return
  for Venta in ventas:
    Venta.mostrar_info()

def mostrar_menu():
  print("\n --- MENU PRINCIPAL --- ")
  print("1. Dar de alta un producto")
  print("2. Mostrar inventario de productos")
  print("3. Consultar producto por codigo")
  print("4. Registrar cliente")
  print("5. Consultar cliente por ID")
  print("6. Mostrar todos los clientes")
  print("7. Registrar una venta")
  print("8. Añadir stock a un producto existente")
  print("9. Mostrar historial de ventas")
  print("10. Salir")

def iniciar_sistema():
  while True:
    mostrar_menu()
    opcion = input("Seleccione una opcion: ")

    if opcion == "1":
      alta_producto()
    elif opcion == "2":
      Mostrar_Inventario()
    elif opcion == "3":
      Consultar_Producto()
    elif opcion == "4":
      Registrar_Cliente()
    elif opcion == "5":
      Consultar_Cliente()
    elif opcion == "6":
      Mostrar_Clientes()
    elif opcion == "7":
      Registrar_Venta()
    elif opcion == "8":
      Añadir_Inventario()
    elif opcion == "9":
      historial_ventas()
    elif opcion == "10":
      print("Gracias por usar el sistema.")
      break
    else:
      print("Opcion invalida. Intente de nuevo.")

### Iniciar el sistema ###

productos = []
clientes = []
ventas = []
inventarios = []

iniciar_sistema()
```
