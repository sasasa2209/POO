```python
import time
import random

# --- SIMULACIÓN DE DATOS (Sustituye a la librería RPi.GPIO) ---
# Definimos valores constantes como lo haría la librería real
BCM = "Modo BCM"
IN = "Entrada"

def gpio_input(pin):
    """Simula la lectura física de un pin devolviendo 0 o 1 al azar"""
    return random.randint(0, 1)

# --- CONFIGURACIÓN (Variables Globales) ---
pin_sensor = 17
print(f"Configurando el pin {pin_sensor} como {IN} en modo {BCM}...")

# --- FUNCIÓN DE LÓGICA ---
def medir_humedad():
    # Paso 1: Obtener el valor crudo del "sensor"
    valor = gpio_input(pin_sensor)

    # Paso 2: Interpretar el valor (Lógica de control)
    if valor == 0:
        estado = "Humedo"
    else:
        estado = "Seco"

    # Paso 3: Mostrar resultado
    print(f"Lectura del sensor: {estado}")

# --- PROGRAMA PRINCIPAL ---
print("--- Iniciando Monitoreo Estructurado ---")

# Ciclo infinito (limitado a 5 para prueba)
for i in range(5):
    medir_humedad()
    time.sleep(2)

print("Fin de la simulación.")
```
